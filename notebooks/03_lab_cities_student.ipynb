## Exercise 4 - Wideband channel

The radio maps used so far give only the power of each pixel: a narrowband prediction. For a
wideband signal, the arrival times of the paths matter too. Path $i$ reaches a receiver with a
complex amplitude $a_i$ after a delay $\tau_i$; the RMS delay spread $\tau_\mathrm{rms}$ measures how
widely the power is spread in time around the mean delay $\bar\tau$:
$$\bar\tau=\frac{\sum_i|a_i|^2\tau_i}{\sum_i|a_i|^2},\qquad
\tau_\mathrm{rms}=\sqrt{\frac{\sum_i|a_i|^2\tau_i^2}{\sum_i|a_i|^2}-\bar\tau^2}.$$
The coherence bandwidth $B_C$ is the band over which the channel stays roughly flat: a signal wider
than $B_C$ is distorted. The larger the delay spread, the smaller $B_C$. This lab estimates $B_C$ with
the rule of thumb $B_C\approx1/(5\tau_\mathrm{rms})$: an order of magnitude, not an exact relation.
For comparison, 3GPP UMi gives median delay spreads of 40 ns in LoS and 97 ns in NLoS at 3.5 GHz.

To get the paths, the code below places 400 receivers at random in the studied area (outdoors, 20 to
250 m from the TX) and computes their paths with `PathSolver`. The code then plots the CDF of
$\tau_\mathrm{rms}$ in LoS and in NLoS for each city, with the UMi medians as reference lines, and adds
the medians, $B_C$ and the shares of receivers in LoS and without paths to the table. The medians
change by a few % between runs and between computers: only larger differences are meaningful.
The last part of the exercise checks how many receivers are needed for reliable curves and medians.

The exercise works in four steps:
1. **Candidates.** The first cell puts a small cell (10 m high, 30 dBm) at every outdoor point of a grid

   one run. With $10^6$ rays per candidate this is a quick screening: good enough to compare the

3. **Check.** The code computes the radio map of the three chosen sites again, with $10^7$ rays, and adds

   at all; and the interference. The code also shows the map of the three sites.
4. **Channel.** The code serves each of the 400 receivers of Exercise 4 from its best site and compares
   the delay spreads with those from the small cell on the square.

At the same total power, two sites at 30 dBm radiate as much as one at 33 dBm, and three as one at
34.8 dBm: so the best site with +3 dB is compared with two sites, and with +4.8 dB with three. With
several sites, each pixel is served by the site it receives best, its **best server**. On one shared
carrier the other sites interfere: the signal-to-interference ratio (SIR) of a pixel is the power of its
best server divided by the sum of the powers of the other sites. The table gives the share of the
covered pixels with an SIR below 6 dB (strong interference).

**Your code** (ref helpers 1.1, 1.5, 2.5): complete two functions. Both receive `prx_all`, the received
power in dBm of every candidate in every pixel (shape [candidates, rows, columns]), and `area`, the mask
of the studied area (True for its pixels, shape [rows, columns]). A layout is a list of candidate
numbers (indices along the first axis of `prx_all`), e.g. `[4, 17]`.
- `layout_coverage(prx_all, sites, area, extra_db)` returns the coverage (%) of `area` with the layout
  `sites`: each pixel gets the power of its best server plus `extra_db` dB (the same extra power at
  every site), and is covered if this power reaches $P_{R,\min}$.
- `greedy_sites(prx_all, area, num_sites)` chooses the sites one at a time: it starts with an empty
  layout and, `num_sites` times, adds the candidate that gives the largest coverage together with the
  sites already chosen. It returns the layout, a list of `num_sites` candidate numbers, the best site
  first.
